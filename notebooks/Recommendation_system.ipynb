```python
import nltk
from nltk import WordNetLemmatizer
from nltk.corpus import stopwords
import re

nltk.download('stopwords')
nltk.download('wordnet')

def preprocess_text(text):
  text = text.lower()
  text = re.sub(r'[^a-zA-Z0-9\s]', '', text)


  stop_words = set(stopwords.words('english'))
  wrodnet = WordNetLemmatizer()

  words = text.split()
  words = [word for word in words if word not in stop_words]
  words = [wrodnet.lemmatize(word) for word in words]

  return " ".join(words)

```

```python
df['tags'] = df['tags'].apply(preprocess_text)
```

```python
df.reset_index(drop=True,inplace=True)
```

```python
indacies = pd.Series(df.index,index = df['title'])
indacies
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer(max_features = 50000,ngram_range=(1,2),stop_words='english')
tfidf_metrix = vectorizer.fit_transform(df['tags'])
```

```python
from sklearn.metrics.pairwise import cosine_similarity

def reccomedation(title,n = 10):
  if title not in indacies:
    return "Movie not found"
  idx = indacies[title]
  sim_scores = list(enumerate(cosine_similarity(tfidf_metrix[idx],tfidf_metrix)[0]))
  sim_scores = sorted(sim_scores,key = lambda x:x[1],reverse=True)

  sim_scores = sim_scores[1:n+1]
  movie_indices = [i[0] for i in sim_scores]
  return df['title'].iloc[movie_indices]
```

```python
reccomedation('Father of the Bride Part II')
```
